```python
import pandas as pd
import os

mimic_path = "/hpc/home/yy450/link_kamaleswaranlab/mimic_iv/builtdata/csv_exports"
output_path = "/hpc/home/yy450/link_kamaleswaranlab/mimic_iv/mimic_flat_files"

os.makedirs(output_path, exist_ok=True)
```

```python
micro_df = pd.read_csv(os.path.join(mimic_path, "hosp_microbiologyevents.csv"))
micro_df.head()
```

```python
cultures_df = micro_df.rename(columns={
    "subject_id": "pat_id",
    "hadm_id": "csn",
    "charttime": "specimen_collect_time",
    "chartdate": "order_time",
    "storetime": "lab_result_time",
    "test_itemid": "proc_code",
    "test_name": "proc_desc",
    "spec_itemid": "component_id",
    "spec_type_desc": "component"
})

cultures_df["loinc_code"] = "NOT AVAILABLE"

cultures_df["result_status"] = "Not Recorded"
```

```python
final_cols = [
    "csn",
    "pat_id",
    "specimen_collect_time",
    "order_time",
    "lab_result_time",
    "result_status",
    "proc_code",
    "proc_desc",
    "component_id",
    "component",
    "loinc_code",
    "org_name",
    "ab_name",
]

cultures_final = cultures_df[final_cols]
cultures_final.head(5)
```

```python
output_file = os.path.join(output_path, "CULTURES.csv")
cultures_final.to_csv(output_file, index=False)

print(f"✅ CULTURES file created: {output_file}")
```
