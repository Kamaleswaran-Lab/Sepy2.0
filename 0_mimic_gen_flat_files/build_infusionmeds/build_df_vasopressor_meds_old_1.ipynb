```python
import pandas as pd

# Step 1: Load the vasoactive agent CSV
vaso_path = "/hpc/home/yy450/link_kamaleswaranlab/mimic_iv/builtdata/csv_concepts_exports/vasoactive_agent.csv"
df = pd.read_csv(vaso_path)

# Step 2: Drop 'endtime' and 'milrinone' columns
df = df.drop(columns=['endtime', 'milrinone'])

# Step 3: Add *_dose_unit columns for each remaining drug
drug_cols = ['dopamine', 'epinephrine', 'norepinephrine', 'phenylephrine', 'vasopressin', 'dobutamine']
for drug in drug_cols:
    df[f"{drug}_dose_unit"] = "mcg/kg/min"
    
df[f"vasopressin_dose_unit"] = "units/hour"

# Step 4: Load ICU stay data and merge on stay_id
icu_path = "/hpc/home/yy450/link_kamaleswaranlab/mimic_iv/builtdata/csv_exports/icu_icustays.csv"
icu_df = pd.read_csv(icu_path, usecols=['stay_id', 'subject_id', 'hadm_id'])

df_merged = df.merge(icu_df, on="stay_id", how="left")

# Step 5: Rename columns
df_merged = df_merged.rename(columns={
    'starttime': 'med_order_time',
    'subject_id': 'pat_id',
    'hadm_id': 'csn'
})

# Step 6: Save to CSV
output_path = "/hpc/home/yy450/link_kamaleswaranlab/mimic_iv/mimic_flat_files/INFUSIONMEDS/df_vasopressor_meds.csv"
df_merged.to_csv(output_path, index=False)

print("Done. Saved to", output_path)
```
