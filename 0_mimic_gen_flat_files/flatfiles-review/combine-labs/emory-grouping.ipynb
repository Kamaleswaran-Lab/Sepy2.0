# Emory's grouping file doesn't have proc_cat_name and proc_desc. This script is to add them back to the grouping file for plotting the histograms in order to compare with MIMIC.

```python
import pandas as pd
import sys

# Read file paths and join column from user input
path1 = '/hpc/home/yy450/link_dctrl_yy450/Sepy2.0/groupings/em_grouping_labs.csv'
path2 = '/hpc/home/yy450/link_kamaleswaranlab/EmoryDataset/EMR_RAW/2022/CJSEPSIS_LABS_2022.dsv'
join_col = 'component_id'
output_path = '/hpc/home/yy450/link_dctrl_yy450/Sepy2.0/0_mimic_gen_flat_files/flatfiles-review/combine-labs/emory-labs-grouping-joined.csv'

# Read CSV files
df1 = pd.read_csv(path1)
df2 = pd.read_csv(path2, sep='|')

df2 = df2[[join_col, "proc_cat_name", "proc_desc"]]
df2 = df2.drop_duplicates(subset=[join_col])

# Join on the specified column
joined_df = pd.merge(df1, df2, on=join_col, how='left')

# Write the result to a new CSV
joined_df.to_csv(output_path, index=False)

print(f"Joined file saved to {output_path}")
```
