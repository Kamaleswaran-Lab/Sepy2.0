```python
import pandas as pd

# -------- USER INPUT --------
file1 = "/hpc/home/yy450/link_dctrl_yy450/Sepy2.0/mimic_groupings/mimic_grouping_labs_new.csv"
file2 = "/hpc/home/yy450/link_dctrl_yy450/Sepy2.0/groupings/em_grouping_labs.csv"
target_column = "super_table_col_name"
# ----------------------------

def load_and_filter(file_path):
    df = pd.read_csv(file_path)
    print(f"\n✅ Loaded file: {file_path} ({df.shape[0]} rows, {df.shape[1]} columns)")

    if "import" not in df.columns:
        raise ValueError(f"'import' column not found in file: {file_path}")
    if target_column not in df.columns:
        raise ValueError(f"'{target_column}' not found in file: {file_path}")
    
    df_filtered = df[df["import"] == "Yes"]
    print(f"🔍 Rows with import == 'Yes': {df_filtered.shape[0]}")
    
    values = df_filtered[target_column].dropna().unique()
    print(f"📋 Unique '{target_column}' values ({len(values)}):")
    print(sorted(values))
    return set(values)

# Load both datasets
values1 = load_and_filter(file1)
values2 = load_and_filter(file2)

# Compare differences
only_in_file1 = values1 - values2
only_in_file2 = values2 - values1
in_both = values1 & values2

print("\n📌 Differences between the two files:")
print(f"✅ Present only in file1 ({len(only_in_file1)}):")
print(sorted(only_in_file1))

print(f"\n✅ Present only in file2 ({len(only_in_file2)}):")
print(sorted(only_in_file2))

print(f"\n🤝 Present in BOTH files ({len(in_both)}):")
print(sorted(in_both))
```
