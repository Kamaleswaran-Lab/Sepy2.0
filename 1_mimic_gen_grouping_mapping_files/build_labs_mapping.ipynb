```python
import pandas as pd

mapping_file = "/hpc/home/yy450/link_dctrl_yy450/Sepy2.0/mimic_groupings/mimic_mapping_labs.csv"
labitems_file = "/hpc/home/yy450/link_kamaleswaranlab/mimic_iv/builtdata/csv_exports/hosp_d_labitems.csv"
output_file = "/hpc/home/yy450/link_dctrl_yy450/Sepy2.0/mimic_groupings/mimic_mapping_labs_with_id.csv"

mapping_df = pd.read_csv(mapping_file)

labitems_df = pd.read_csv(labitems_file, usecols=["itemid", "label"])

merged = mapping_df.merge(
    labitems_df, left_on="component", right_on="label", how="left"
)

merged = merged.rename(columns={"itemid": "component_id"}).drop(columns=["label"])

merged["component_id"] = merged["component_id"].astype("Int64")

merged.to_csv(output_file, index=False)

print(f"✅ New file has been saved to {output_file}, total {merged.shape[0]} rows, {merged.shape[1]} columns")
print("The first 5 rows are:")
display(merged.head())
```
