```python
import pandas as pd

# 读入 CSV
df = pd.read_csv('/hpc/home/yy450/link_kamaleswaranlab/mimic_iv/builtdata/csv_exports/hosp_transfers.csv')

# 找出 outtime 为空的行
null_outtime_rows = df[df['outtime'].isna()]

# 提取这些行的 eventtype 列的唯一值
unique_eventtypes = null_outtime_rows['eventtype'].unique()

# 打印结果
print("outtime 为空的行对应的 eventtype 取值如下：")
for val in unique_eventtypes:
    print(val)

# 用法示例
# check_outtime_null_eventtypes("/path/to/your/file.csv")
```

```python
import pandas as pd

# 输入文件路径
input_file = "/hpc/home/yy450/link_kamaleswaranlab/mimic_iv/mimic_flat_files/VENT.csv"

# 要检查的两列名称
col1 = "vent_start_time"
col2 = "vent_stop_time"

# 代表病例号的列名
csn_col = "csn"

# 读取 CSV
df = pd.read_csv(input_file)

# 筛选出 col1 和 col2 都为空的行
# mask = df[col1].isna() & df[col2].isna()
mask = df[col2].isna()
empty_rows = df[mask]

# 输出对应的 csn 值
csn_values = empty_rows[csn_col].unique()

print("以下 csn 对应的行在两列都为空：")
for v in csn_values:
    print(v)
```
