```python
import pickle
import pandas as pd

# 1. Set file path
file_path = "/hpc/home/yy450/link_kamaleswaranlab/mimic_iv/sepy_output/mimic-supertables/ClinicalData/20093378.pkl"
output_file = "mimic_pickle_contents_summary.txt"

# 2. Read pickle
data = pickle.load(open(file_path, "rb"))

# 3. Open txt file and write contents
with open(output_file, "w", encoding="utf-8") as f:
    f.write(f"Contents of pickle file: {file_path}\n")
    for key, value in data.items():
        f.write(f"\n==== {key} ({type(value)}) ====\n")

        if isinstance(value, pd.DataFrame):
            f.write(f"DataFrame Size: {value.shape}\n")
            f.write(f"Column Names: {value.columns.tolist()}\n")
            f.write("Head (first 5 rows):\n")
            f.write(value.head().to_string() + "\n")

        elif isinstance(value, pd.Series):
            f.write(f"Series Length: {len(value)}\n")
            f.write("Head (first 5 rows):\n")
            f.write(value.head().to_string() + "\n")

        elif isinstance(value, dict):
            f.write(f"Number of Keys: {len(value)}\n")
            for i, (k, v) in enumerate(value.items()):
                f.write(f"{k}: {v}\n")
                if i >= 9:  # Only write the first 10 key-value pairs
                    f.write("... (truncated)\n")
                    break

        elif isinstance(value, (list, tuple)):
            f.write(f"Length: {len(value)}\n")
            f.write(f"First 10 items: {value[:10]}\n")

        else:
            f.write(str(value) + "\n")

print(f"✅ Output complete, contents written to {output_file}")
```
